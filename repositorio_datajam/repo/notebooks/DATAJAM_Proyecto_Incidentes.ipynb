## 14. Limpieza de datos: Eliminación de columnas no relevantes o con altos valores nulos
En esta sección depuraremos el dataset eliminando aquellas columnas que no aportan valor a nuestra pregunta analítica o que tienen una cantidad excesiva de datos faltantes, optimizando así el modelo.

```python
columnas_a_eliminar = [
    'Hombres Sin Signos'
    'Mujeres Sin Signos'
    'Menores Niñas Sin Signos'
    'Menores Niñños Sin Signos'
]

df.drop(columns=columnas_a_eliminar, inplace=True, errors='ignore')
```

## 15. Construcción del perfil general del conjunto de datos

```python
print(f"Total registros: {len(df)}")
print(f"Total columnas: {len(df.columns)}")
print(f"cantidad total de nulos: {df.isnull().sum().sum()}")
print(f"Registros duplicados: {df.duplicated().sum()}")
print(f"Limpieza completada. El Dataset ahora tien {df.shape[0]} filas y {df.shape[1]} columnas.")
```
